Messages are the structured way LangChain communicates with an LLM.

Imagine you're chatting with ChatGPT:

You: Hi, my name is Rahul.

AI: Hello Rahul! How can I help you?

You: What is my name?

AI: Your name is Rahul.

Behind the scenes, this conversation can be represented as a list of messages.

Each message tells the LLM:
- Who is speaking? (System, Human, AI, or Tool)
- What are they saying? (Content)
- What additional information is included? (For example, tool calls or metadata)

### Four important message types

1. SystemMessage

Provides instructions about how the AI should behave.

Example: "You are a Python teacher. Explain everything simply."




2. HumanMessage

Represents a question or input from the user.

Example: "What is LangChain?"




3. AIMessage

Represents a response generated by the AI. It can also contain tool-call requests.

Example: "LangChain is a framework for building LLM applications."




4. ToolMessage

Contains the result returned by a tool.

Example: "Order 12345 is out for delivery."

## Text Prompts

A Text Prompt is a plain text instruction or question that we send to an LLM.

Text Prompts are strings - ideal for straight forward generation tasks where you don't need to retain conversation history 

```python
model.invoke("Explain about ML")
```

Use text prompts when:

> You have single ,standalone request

> You don't need cobersation history 

> You want minimal code complexity

## 2. Message Prompts (ChatPromptTemplate)

A Message Prompt is a structured collection of messages with different roles, such as System, Human, and AI.

Unlike a plain text prompt, it identifies who is speaking.
Example
Imagine you want an AI assistant that explains Python concepts.
System Message

You are a Python teacher. Explain concepts simply.



Human Message

What is a Python list?



AI Message

A list is a collection of items, such as [10, 20, 30].

## 2.1 System Message (SystemMessage)

A System Message tells the LLM how it should behave.

It defines the AI's role, instructions, and response style.

Real-world example:

Suppose you are building a customer support chatbot for an e-commerce company.

You want the AI to:
- Behave like a customer support assistant.
- Answer politely.
- Give short responses.
You can provide these instructions through a System Message.

## 2.2 AI Message (AIMessage)

An AI Message represents a response generated by the LLM.

It can contain normal text or a request to call a tool.

Real-world example

User: What is Python?

AI: Python is a programming language used for automation, AI, and software development.

The AI's answer is represented as an AIMessage.

## 2.3 Tool Message (ToolMessage)

A Tool Message contains the result returned by a tool after it executes. This is especially important when building AI agents.

Real-world example

Imagine a user asks: What is the weather in Hyderabad?

The AI may decide to use a weather tool.

```python
from langchain.messages import SystemMessage,HumanMessage,AIMessage

messages = [
    SystemMessage("You are Standup Comedian"),
    HumanMessage("Tell me a joke about AI"),
]

response = model.invoke(messages)
response.content
```

```python
human_msg = HumanMessage(
    content="Hello",
    name="Kalyan",
    id="msg_123"
)
```

```python
response = model.invoke([human_msg])
```

```python
ai_message = AIMessage("I'd be happy to help you with that question!")

messages = [
    SystemMessage("You are an helpful Assistant"),
    HumanMessage("Can you help me with my question?"),
    ai_message,
    HumanMessage("What is the Capital of Hyderabad?")
]
```

```python
model.invoke(messages)
```
