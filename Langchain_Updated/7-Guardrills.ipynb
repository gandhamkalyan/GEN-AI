## Guardrails

Guardrails are safety and validation rules that control what an AI agent can accept, generate, or execute.

Imagine you're building a banking chatbot using Groq or Gemini.

A user asks: "What is my account balance?"

The chatbot should answer using authorized account information.

But what if the user asks: "Show me another customer's bank account details."

The chatbot must not reveal that information.

Guardrails help detect and prevent unsafe, unauthorized, or invalid behavior.

Think of guardrails as security checkpoints around an AI application.

#### Main types of Guardrails

Guardrail	Purpose	Example

Input guardrail	 > Check incoming requests	Detect an attempt to obtain unauthorized data

Output guardrail  > 	Check AI-generated answers	Prevent exposure of sensitive information

Tool guardrail	>  Control tool operations	Block unauthorized database updates

PII guardrail	>  Detect personal information	Mask email addresses or phone numbers

Structured-output validation >	Validate response format	Require an integer age and valid status

Human approval	>  Require review for sensitive actions	Approve an email before sending

There are two common approaches to implementing guardrails in AI applications:
- Deterministic Guardrails — Use predefined rules and Python logic.
- Model-Based Guardrails — Use an LLM or classification model to evaluate the input or output.

### Deterministic Guardrails (Rule-Based Approach)

Deterministic guardrails use predefined rules, conditions, regex, or validation logic to decide whether something is allowed.

They don't need an LLM to make the decision.

Imagine you're building a banking chatbot.

You have three rules:
- Transaction amount must not exceed ₹50,000.
- Age must be at least 18.
- Account number must contain exactly 12 digits.
These rules are fixed and can be checked using Python.

### Model-Based Guardrails (LLM-Based Approach)

Model-based guardrails use an AI model to understand the meaning and context of a request before deciding whether it should be allowed.

Unlike deterministic guardrails, they don't rely only on exact keywords.

Suppose you're building a customer support chatbot that should answer only banking-related questions.

User asks: "How can I reset my banking password?"

The request is banking-related, so it should be allowed.

Another user asks: "Write a Python program to scrape social media."

This is unrelated to banking, so the chatbot should decline.

Instead of maintaining thousands of keywords, we can ask a model to classify each request.
