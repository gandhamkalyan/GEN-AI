## Streaming

> Streaming means receiving the AI's response gradually instead of waiting for the entire response to be generated.

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
model_gem.invoke("Write a poem about the beauty of nature in the style of William Wordsworth.")
```

```python
model_gem.stream("Write a poem about the beauty of nature in the style of William Wordsworth.")
```

```python
for chunk in model_gem.stream("Explain about AI."):
    print(chunk.text, end="|",flush=True)
```

## Batch

Batching means processing multiple inputs together instead of calling the LLM separately for each input in your application code.

Imagine you have five questions:
1. What is AI?
2. What is Machine Learning?
3. What is Deep Learning?
4. What is NLP?
5. What is Generative AI?
Normally, you might process them one after another.

With batching, you submit all five inputs through a single LangChain method, and LangChain manages their execution.

```python
responses = model_gr.batch([
    "What is AI",
    "Explain about DL",
    "What are the best LLMs to use for RAG"
],
    config={"max_concurrency": 5}  #limit to 5 parallel requests
)
```

```python
for response in responses:
    print(response)
```
